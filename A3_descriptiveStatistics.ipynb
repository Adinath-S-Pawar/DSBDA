```python
import pandas as pd
from sklearn.datasets import load_iris

# Load iris dataset from sklearn
iris = load_iris()

# Create DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Add species column
df['species'] = [iris.target_names[i] for i in iris.target]
```

```python
# Display first 5 rows
print("First 5 Rows:\n")
df.head()
```

```python
print("\n\nSUMMARY STATISTICS GROUPED BY SPECIES\n")
# categorial value = species, quantatative variables = sepal and petal len,width
summary = df.groupby('species').agg(
    ['mean', 'median', 'min', 'max', 'std']
).round(2)

summary
```

```python
# 2 Statistical Details For Each Species
species_list = df['species'].unique()

for species in species_list:

    print("\n===================================")
    print("Species :", species)
    print("===================================")

    # Filter rows of one species
    data = df[df['species'] == species]

    # Basic statistical details
    print("\nBasic Statistics:\n")
    display(data.describe().round(2))

    # Percentiles
    print("\nPercentiles:\n")
    display(
        data.quantile(
            [0.25, 0.50, 0.75],
            numeric_only=True
        ).round(2)
    )
```
