```python
import seaborn as sns #statistical data visualization library

df = sns.load_dataset('titanic') #Seaborn’s built-in Titanic dataset
df.head()
```

```python
sns.boxplot(x='sex',y='age',data=df)
```

```python
sns.boxplot(x='sex', y='age', data=df, hue='survived')
```

```python
sns.catplot(x='survived',data=df, kind='count',hue='sex') #A high-level Seaborn function for plotting categorical data.
```

```python
sns.catplot(data=df, x="age", y="class") #default kind = "strip"
```
