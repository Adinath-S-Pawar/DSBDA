```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic Dataset
dataset = sns.load_dataset('titanic')

# Display First 5 Rows
print("FIRST 5 ROWS\n")
display(dataset.head())
```

```python
#find patterns in Titanic dataset

# Dataset Information
print("\nDATASET INFORMATION\n")
print(dataset.info())

# Statistical Summary
print("\nSTATISTICAL SUMMARY\n")
display(dataset.describe())
```

```python
# Countplot of Survival
# Shows number of survived and non-survived passengers
plt.figure(figsize=(6,4))
sns.countplot(x='survived', data=dataset)
plt.title("Survival Count")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='sex', hue='survived', data=dataset) #separate data using different colors based on another category
plt.title("Survival based on Gender")
plt.show()
```

```python
# Survival based on Passenger Class
# Shows survival distribution for each passenger class
plt.figure(figsize=(6,4))
sns.countplot(x='pclass', hue='survived', data=dataset)
plt.title("Survival based on Passenger Class")
plt.show()
```

```python
# Boxplot of Fare by Passenger Class
# Shows fare distribution and outliers for each class
plt.figure(figsize=(8,5))
sns.boxplot(x='pclass', y='fare', data=dataset)
plt.title("Fare Distribution by Passenger Class")
plt.show()
```

```python
# Histogram of Fare
# Shows distribution of ticket prices
plt.figure(figsize=(8,5))
sns.histplot(dataset['fare'], bins=30)
plt.title("Distribution of Titanic Fare")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
plt.show()
```
